```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data=pd.read_csv("UK_Accident.csv",index_col=0)
```

```python
data.head()
```

```python
data.drop("Accident_Index",axis=1,inplace=True)
```

```python
data.head()
```

```python
total_row=data.shape[0]
```

```python
data_null=data.isnull().sum()[data.isnull().sum() > 0]
```

```python
data_null
```

```python
data_null_percentage=(data_null/total_row)*100
```

```python
data_null_percentage
```

```python
data['Junction_Control'].value_counts()
```

```python
data['Special_Conditions_at_Site'].value_counts()
```

```python
data['Carriageway_Hazards'].value_counts()
```

```python
data['LSOA_of_Accident_Location'].value_counts()
```

```python
Colums_to_drop=data_null_percentage[data_null_percentage>30].index
```

```python
Colums_to_drop
```

```python
new_data=data.drop(Colums_to_drop,axis=1)
new_data.dropna(inplace=True)
```

```python
new_data.head()
```

```python
new_data.columns[new_data.isnull().sum()>0]
```

```python
new_data.isnull().sum()
```

```python
percent_of_data_remaining_row=(new_data.shape[0]/data.shape[0])*100
```

```python
percent_of_data_remaining_column=(new_data.shape[1]/data.shape[1])*100
```

```python
print(f"Row Percentage of data remaining after cleaning: {percent_of_data_remaining_row:.2f}%")
print(f"Column Percentage of data remaining after cleaning: {percent_of_data_remaining_column:.2f}%")
```

```python
new_data.sort_values('Date', ascending=True, inplace=True)
```

```python
new_data['Date'].info
```

```python
new_data["Date"] = pd.to_datetime(new_data["Date"], dayfirst=True)  # skip if already datetime
new_data["Month"] = new_data["Date"].dt.month  
new_data["Month_Name"] = new_data["Date"].dt.month_name()
new_data["Quarter"] = new_data["Date"].dt.quarter
```
