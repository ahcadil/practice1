```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import matplotlib.ticker as ticker
```

```python
data_load=pd.read_feather('output_file_name.feather')
```

```python
new_data = data_load.reset_index(drop=True)
new_data.insert(0, "Index", range(1, len(new_data) + 1))
```

```python
new_data["Road_Type"].value_counts()
```

```python
counts = new_data['Quarter'].value_counts().sort_index()
plt.bar(counts.index, counts.values, color="orange")
plt.xlabel("Quarter"); plt.ylabel("Number of accidents")
plt.xticks([1, 2, 3, 4], ["Q1", "Q2", "Q3", "Q4"])
plt.show()
```

```python
counts_month = new_data["Month"].value_counts().sort_index()
plt.figure(figsize=(15, 7),)
ax=sns.barplot(x=counts_month.index, y=counts_month.values,color="darkgray")
ax.set_xticklabels(["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"])
plt.show()
```

```python


plt.figure(figsize=(16, 7))

# Added width and a subtle gap so the 7 daily bars per year are distinct
ax = sns.countplot(
    data=new_data,
    x="Year",
    hue="Day_of_Week",
    palette="Paired",
    edgecolor="black",
    width=0.85,
    gap=0.05
)

# Safely map custom day labels to the existing legend handles
handles, _ = ax.get_legend_handles_labels()
ax.legend(
    handles=handles,
    title="Day of Week",
    labels=["Sunday", "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday"],
    loc="upper left",
    bbox_to_anchor=(1.01, 1),
    frameon=False,
    title_fontsize=12,
    fontsize=11
)

# Added main chart title and styled both axis labels
ax.set_title("Accidents by Year and Day of Week", fontsize=16, fontweight="bold", pad=15)
ax.set_xlabel("Year", fontsize=14, labelpad=10)
ax.set_ylabel("Count", fontsize=14, labelpad=10)

# Styled tick labels and formatted large y-axis numbers with commas
plt.xticks(fontsize=11)
plt.yticks(fontsize=11)
ax.yaxis.set_major_formatter(ticker.StrMethodFormatter("{x:,.0f}"))

# Removed top and right borders and prevented legend clipping
sns.despine()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15, 7))

# Increased width to 0.75 so the 4 quarterly bars per year have balanced proportions
ax = sns.countplot(
    data=new_data,
    x="Year",
    hue="Quarter",
    palette="Paired",
    edgecolor="black",
    width=0.75,
    gap=0.1
)

# Safely map custom quarter labels to the existing legend handles
handles, _ = ax.get_legend_handles_labels()
ax.legend(
    handles=handles,
    title="Quarter",
    labels=["Q1", "Q2", "Q3", "Q4"],
    loc="upper left",
    bbox_to_anchor=(1.01, 1),
    frameon=False,
    title_fontsize=12,
    fontsize=11
)

# Added main chart title and styled both axis labels
ax.set_title("Accidents by Year and Quarter", fontsize=16, fontweight="bold", pad=15)
ax.set_xlabel("Year", fontsize=14, labelpad=10)
ax.set_ylabel("Count", fontsize=14, labelpad=10)

# Styled tick labels and formatted large y-axis numbers with commas
plt.xticks(fontsize=11)
plt.yticks(fontsize=11)
ax.yaxis.set_major_formatter(ticker.StrMethodFormatter("{x:,.0f}"))

# Removed top and right borders and prevented legend clipping
sns.despine()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15, 7), dpi=100)

# Adjusted width to 0.65 for clean proportions with 2 bars per day
ax = sns.countplot(
    data=new_data,
    x="Day_of_Week",
    hue="Urban_or_Rural_Area",
    palette="Paired",
    edgecolor="black",
    width=0.65,
    gap=0.1
)

# Safely map custom "Urban" and "Rural" labels to the existing legend handles
handles, _ = ax.get_legend_handles_labels()
ax.legend(
    handles=handles,
    title="Area of Accident",
    labels=["Urban", "Rural"],
    loc="upper left",
    bbox_to_anchor=(1.01, 1),
    frameon=False,
    title_fontsize=12,
    fontsize=11
)

# Added main chart title and styled both axis labels
ax.set_title("Accidents by Day of Week and Area (Urban vs. Rural)", fontsize=16, fontweight="bold", pad=15)
ax.set_xlabel("Day of Week", fontsize=14, labelpad=10)
ax.set_ylabel("Count", fontsize=14, labelpad=10)

# Replaced numeric day ticks with readable day names and styled tick labels
plt.xticks(
    ticks=range(7),
    labels=["Sunday", "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday"],
    fontsize=11
)
plt.yticks(fontsize=11)

# Formatted large y-axis numbers with commas
ax.yaxis.set_major_formatter(ticker.StrMethodFormatter("{x:,.0f}"))

# Removed top and right borders and prevented legend clipping
sns.despine()
plt.tight_layout()
plt.show()
```
